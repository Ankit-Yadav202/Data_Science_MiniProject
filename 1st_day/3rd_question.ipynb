```python
from statistics import *

# 1st function
def fun1(lst):

    while len(lst) < 9:
        lst.append(None)

    return [lst[0:3], lst[3:6], lst[6:9]]

# 2nd function
def fun2(lst):

    t = tuple(lst)

    return {
        "Tuple": t,
        "Mean": mean(lst),
        "Median": median(lst),
        "Mode": mode(lst)
    }

# 3rd function
def fun3(lst):

    return {
        "Function1": fun1(lst.copy()),
        "Function2": fun2(lst.copy())
    }

lst = list(map(int, input("Enter numbers: ").split()))

print(fun3(lst))
```
